```python
import pandas as pd

# CSV 파일 불러오기
csv_filename = "./review_by_gpt.csv"
df = pd.read_csv(csv_filename, encoding="utf-8-sig")

# 데이터 확인
print(df.head())
```

```python
import re

def clean_text(text):
    text = text.lower()  # 소문자 변환
    text = re.sub(r"[^가-힣a-zA-Z0-9\s]", "", text)  # 한글, 영어, 숫자만 남기기
    text = re.sub(r"\s+", " ", text).strip()  # 연속된 공백 제거
    return text

df["리뷰"] = df["리뷰"].apply(clean_text)
```

```python
# 데이터 확인
print(df.head())
```

# 데이터 전처리, 토큰화

```python
from soynlp.tokenizer import RegexTokenizer

tokenizer = RegexTokenizer()
tokens = tokenizer.tokenize("이 편의점 신상품 너무 맛있어요!")
print(tokens)
```

```python
from soynlp.tokenizer import RegexTokenizer

# ✅ soynlp 토큰화 수행
tokenizer = RegexTokenizer()
df["토큰화된 리뷰"] = df["리뷰"].apply(tokenizer.tokenize)

# ✅ 토큰화된 데이터 확인
print(df[["리뷰", "토큰화된 리뷰"]].head())
```

```python
from sklearn.preprocessing import LabelEncoder

# ✅ 속성과 감성을 숫자로 변환
aspect_encoder = LabelEncoder()
sentiment_encoder = LabelEncoder()

df["속성_인코딩"] = aspect_encoder.fit_transform(df["속성"])
df["감성_인코딩"] = sentiment_encoder.fit_transform(df["감성"])

# ✅ 변환된 데이터 확인
print(df[["속성", "속성_인코딩", "감성", "감성_인코딩"]].head())
```

```python
from transformers import BertTokenizer

# ✅ BERT 토크나이저 로드
tokenizer = BertTokenizer.from_pretrained("beomi/kcbert-base")

# 최대 토큰 길이 설정
MAX_LEN = 64

def bert_tokenize(text):
    encoding = tokenizer(
        text,
        max_length=MAX_LEN,
        padding="max_length",
        truncation=True,
        return_tensors="pt"
    )
    return encoding["input_ids"][0].tolist(), encoding["attention_mask"][0].tolist()

df["input_ids"], df["attention_mask"] = zip(*df["리뷰"].apply(bert_tokenize))
```

# 모델 학습 및 평가

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split

# 입력 및 레이블 데이터
X = list(zip(df["input_ids"], df["attention_mask"]))
y_aspect = df["속성_인코딩"].values
y_sentiment = df["감성_인코딩"].values

# 데이터 분할
X_train, X_test, y_train_aspect, y_test_aspect, y_train_sentiment, y_test_sentiment = train_test_split(
    X, y_aspect, y_sentiment, test_size=0.2, random_state=42, stratify=y_sentiment
)

print(f"학습 데이터 개수: {len(X_train)}, 테스트 데이터 개수: {len(X_test)}")
```

```python
import torch
from torch.utils.data import DataLoader, TensorDataset

# ✅ PyTorch Tensor 변환 (훈련 데이터)
train_inputs = torch.tensor([x[0] for x in X_train])
train_masks = torch.tensor([x[1] for x in X_train])
train_labels_aspect = torch.tensor(y_train_aspect)
train_labels_sentiment = torch.tensor(y_train_sentiment)

# ✅ PyTorch Tensor 변환 (테스트 데이터)
test_inputs = torch.tensor([x[0] for x in X_test])
test_masks = torch.tensor([x[1] for x in X_test])
test_labels_aspect = torch.tensor(y_test_aspect)
test_labels_sentiment = torch.tensor(y_test_sentiment)

# ✅ DataLoader 생성 (Batch Size = 8)
train_dataset = TensorDataset(train_inputs, train_masks, train_labels_aspect, train_labels_sentiment)
train_dataloader = DataLoader(train_dataset, batch_size=8, shuffle=True)

test_dataset = TensorDataset(test_inputs, test_masks, test_labels_aspect, test_labels_sentiment)
test_dataloader = DataLoader(test_dataset, batch_size=8, shuffle=False)

# ✅ 데이터 크기 확인
train_size = len(train_dataset)
test_size = len(test_dataset)
print(f"훈련 데이터 개수: {train_size}, 테스트 데이터 개수: {test_size}")
```

```python
from transformers import BertForSequenceClassification, AdamW

# ✅ GPU 사용 여부 확인
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ✅ BERT 모델 로드 (출력 레이블 3개: 속성 + 감성)
model = BertForSequenceClassification.from_pretrained("beomi/kcbert-base", num_labels=len(aspect_encoder.classes_) + len(sentiment_encoder.classes_))
model.to(device)

# ✅ 옵티마이저 및 손실 함수 설정
optimizer = AdamW(model.parameters(), lr=2e-5)
loss_fn = torch.nn.CrossEntropyLoss()

# ✅ 학습 실행
EPOCHS = 3

for epoch in range(EPOCHS):
    model.train()
    total_loss = 0

    for batch in train_dataloader:
        batch = tuple(t.to(device) for t in batch)
        inputs, masks, labels_aspect, labels_sentiment = batch

        optimizer.zero_grad()
        outputs = model(inputs, attention_mask=masks)
        loss_aspect = loss_fn(outputs.logits[:, :len(aspect_encoder.classes_)], labels_aspect)
        loss_sentiment = loss_fn(outputs.logits[:, len(aspect_encoder.classes_):], labels_sentiment)

        loss = loss_aspect + loss_sentiment
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
    
    print(f"Epoch {epoch+1}/{EPOCHS}, Loss: {total_loss/len(train_dataloader):.4f}")
```

```python
# ✅ 학습 설정
EPOCHS = 3  # 학습 반복 횟수

for epoch in range(EPOCHS):
    model.train()
    total_loss = 0

    for batch in train_dataloader:
        batch = tuple(t.to(device) for t in batch)  # GPU/CPU 이동
        inputs, masks, labels_aspect, labels_sentiment = batch

        optimizer.zero_grad()
        outputs = model(inputs, attention_mask=masks, labels=labels_sentiment)
        loss = outputs.loss
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
    
    print(f"Epoch {epoch+1}/{EPOCHS}, Loss: {total_loss/len(train_dataloader):.4f}")
```

## 모델 평가 - 인스타그램 크롤링

```python
import pandas as pd
import torch
from transformers import BertTokenizer

# ✅ 새로운 리뷰 데이터 업로드
csv_filename = "./instagram_review.csv"  # 업로드한 파일 이름을 여기에 입력
df_new = pd.read_csv(csv_filename, encoding="utf-8-sig")

# ✅ 리뷰 데이터 확인
print(df_new.head())
```

```python
# ✅ 리뷰 데이터 전처리 (텍스트 정제)
import re

def clean_text(text):
    text = text.lower()  # 소문자로 변환
    text = re.sub(r"[^가-힣a-zA-Z0-9\s]", "", text)  # 한글, 영문, 숫자만 남기기
    text = re.sub(r"\s+", " ", text).strip()  # 연속된 공백 제거
    return text

df_new["리뷰"] = df_new["content"].apply(clean_text)
```

```python
# ✅ 필요한 라이브러리 임포트
import torch
from transformers import BertForSequenceClassification

# ✅ 디바이스 설정 (GPU 사용 가능 여부 확인)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ✅ 저장된 모델 로드
num_labels = len(aspect_encoder.classes_) + len(sentiment_encoder.classes_)  # 속성 + 감성 레이블 개수
model = BertForSequenceClassification.from_pretrained("beomi/kcbert-base", num_labels=num_labels)
model.load_state_dict(torch.load("./absa_model.pth"))
model.to(device)
model.eval()
```

```python
# ✅ BERT 토크나이저 로드
tokenizer = BertTokenizer.from_pretrained("beomi/kcbert-base")

# ✅ 리뷰 데이터를 BERT 입력 형식으로 변환
def bert_tokenize(text):
    if pd.isna(text) or text.strip() == "":  # ✅ 빈 텍스트 처리
        return [0] * 64, [0] * 64  # 패딩 토큰 반환
    
    encoding = tokenizer(
        text,
        max_length=64,
        padding="max_length",
        truncation=True,
        return_tensors="pt"
    )
    return encoding["input_ids"][0].tolist(), encoding["attention_mask"][0].tolist()

df_new["input_ids"], df_new["attention_mask"] = zip(*df_new["content"].apply(bert_tokenize))
```

```python
# ✅ 모델 예측 함수 정의
def predict_sentiment_aspect(input_ids, attention_mask):
    input_ids = torch.tensor(input_ids).unsqueeze(0).to(device)  # 배치 차원 추가
    attention_mask = torch.tensor(attention_mask).unsqueeze(0).to(device)

    with torch.no_grad():
        outputs = model(input_ids, attention_mask=attention_mask)
        logits = outputs.logits

        # ✅ 속성과 감성 예측
        predicted_aspect_label = torch.argmax(logits[:, :len(aspect_encoder.classes_)], dim=1).cpu().item()
        predicted_sentiment_label = torch.argmax(logits[:, len(aspect_encoder.classes_):], dim=1).cpu().item()

        # ✅ 인코딩된 값을 원래 라벨(한글)로 변환
        predicted_aspect = aspect_encoder.inverse_transform([predicted_aspect_label])[0]
        predicted_sentiment = sentiment_encoder.inverse_transform([predicted_sentiment_label])[0]

    return predicted_aspect, predicted_sentiment

# ✅ df_new에 대해 모델 예측 실행
df_new["예측된 속성"], df_new["예측된 감성"] = zip(*df_new.apply(
    lambda row: predict_sentiment_aspect(row["input_ids"], row["attention_mask"]), axis=1))
```
